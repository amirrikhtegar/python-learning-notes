```python
# دستورات کنترلی (Control Flow)

# دو مدل دستور داریم ساده و مرکب 

# import ساده

# if / elif / else
x = 10
if x > 5:
    print("bigger")
elif x == 5:
    print("equal")
else:
    print("smaller")

# while

i = 0
while i < 5:
    print(i)
    i += 1

# for

for i in range(5):
    print(i)

```
