```python
# Arithmetic operators
x = 10
y = 3
z = x + y # اگه اینجوری بنویسم ذخیره میشه در حافظه 
# x + y
# x - y
# x * y
# x / y
# x // y تقسیم فقط با عدد صحیح 
# x % y باقیمانده 
# x ** y توان 

print(x + y) # اگه اینجوری بنویسم ذخیره نمیشه

print(z)

```

```python
# Comparison operators
# از اینا برای operator logical استفاده میشه
x = 23

y = 22

# x == y
# x != y
# x > y
# x < y
# x >= y
# x <= y

print(x == 23)
print(x <= y)

```

```python
# Assignment operators

# x = 5 left side variable right side value
# x += 3 , x = x + 3 
# x -= 3 , x = x - 3
# x *= 3 , x = x * 3
# x /= 3 , x = x / 3
# x %= 3 , x = x % 3
# x //= 3 , x = x // 3
# x **= 3 , x = x ** 3
# x &= 3 , x = x & 3
# x |= 3 , x = x  | 3
# x ^= 3 , x = x ^ 3
# x >>= 3 , x <<= 3 , x = x >>(<<) 3
x = 5
x **= 2
print(x)
x = 5
x = x ** 2
print(x)
```

```python
# Logical operators

x = 3 
y = 3
print(not x > 5 or y >= 3 and x <= 3)
```

```python
# membership operators
# اول چیزی که میخوام معلوم کنم بعد چیزی که میخوام توش جستجو کنم

x = 13

y = [7, 10, 13, 20]

print(x in y)

z = "amir mohamme"

print("mohamme" not in z)
```

```python
# Identity operators

x = [1, 2 ,3]

y = [1, 2 ,3]

# is

# is not

# اکس هست همون وای یا نه

print(x is y)

print(x is not y)

print(x == y)

print(id(x))

print(id(y))
```

```python
# bitwise operators 
# bayad bhtar yad bgerm

x = 15
y = 10
# & , and
# | , or
# ^ , xor
# ~ , not
# << , ZERO FILL LEFT SHIFT
# >> , SIGNAL RIGHT SHIFT

print(bin(y))
print(x & y)
print(x | y)
print(x ^ y)
print(bin(x ^ y))
#print(x << y)
print(bin(x >> y))
# x // 2 ** 1
print(bin(x), bin(y))
print(bin(~y))
# -x -1 
xo = 0b111111111


```

```python
# Walrus operator

# برای اختصاص مقدار به متغیر در داخل یک عبارت استفاده می‌شود

x = 10 

print(x:= x + 4)

print(x)

```
